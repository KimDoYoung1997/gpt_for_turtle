```python
from openai import OpenAI
import os 

api_key = os.getenv('OPENAI_API_KEY')
client = OpenAI(api_key=api_key)
```

```python
completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant. You must answer in Korean."},
        {"role": "user", "content": "대한민국의 수도는 어디인가요"},
    ]
)
print(completion.choices[0].message.content)
```

```python
def ask(question):
    completion=client.chat.completions.create(
        model="gpt-4o-mini",
        messages=[
            {"role": "system", "content": "You are a helpful assistant. You must answer in Korean."},
            {"role": "user", "content": question},
        ]
    )
    return completion.choices[0].message.content    
```

```python
ask("대한민국의 수도는 어디인가요?")
```

```python
ask("영어로 답변해 주세요")
```

# 연속형 대화 구축방법 : Multi-turn

```python
completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant. You must answer in Korean."},
        {"role": "user", "content": "대한민국의 수도는 어디인가요"},    # 1번 질문
        {"role": "assistant", "content": "대한민국 수도는 서울입니다."},# 답변
        {"role": "user", "content": "이전 답변을 영어로 번역해 주세요"}, # 2번 질문
    ]
)
print(completion.choices[0].message.content)
```

```python
def ask(question,message_history=[]):
    if len(message_history)==0:
        message_history.append(
            {"role": "system", 
             "content": "You are a helpful assistant. You must answer in Korean."
             }
        )
    # 사용자 질문 추가
    message_history.append(
        {"role": "user", 
         "content": question})
    # GPT에 질문을 던져 답변을 받음
    completion=client.chat.completions.create(
        model="gpt-4o-mini",
        messages=message_history 
    )
    message_history.append(
        {"role": "assistant", 
         "content": completion.choices[0].message.content}
    )
    print("message_history의 length",len(message_history))
    print("message_history의 id",id(message_history))
    return message_history
```

```python
ask("대한민국의 수도는 어디인가요?")
```

```python
ask("이전 답변 프랑스어로 해")
```

```python
message_history = ask("이전 답변 영어로 해")
```

```python
message_history
```

```python
message_history = ask("이전 답변 사투리로 해",message_history=message_history)
```

# JSON 형식의 답변 활용

```python
completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant designed to output json format."},
        {"role": "user", "content": "대한민국의 수도는 어디인가요"},
    ],
    response_format={"type": "json_object"}
)
print(completion.choices[0].message.content)
```
